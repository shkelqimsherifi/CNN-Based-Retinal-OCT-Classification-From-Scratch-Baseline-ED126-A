# True and Prediction

### Evaluation on Test set

```python
# Suppose you have your model and test data
x_pred = model.predict(test_set)  # Shape: (num_samples, 1)
x_pred_labels = (x_pred > 0.5).astype(int)  # Threshold 0.5

x_true = np.concatenate([y.numpy() for x, y in test_set], axis=0)
```

```python
# Flatten in case labels have shape (num_samples, 1)
x_true_flat = x_true.ravel()      # shape -> (num_samples,)
x_pred_flat = x_pred_labels.ravel()

from sklearn.metrics import confusion_matrix

tn, fp, fn, tp = confusion_matrix(x_true_flat, x_pred_flat).ravel()

print(f"True Negatives: {tn}")
print(f"False Positives: {fp}")
print(f"False Negatives: {fn}")
print(f"True Positives: {tp}")
```

```python
accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0

print(f"Accuracy: {accuracy:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
```

## 🔬 Medical Diagnostic Grid — 4×3 Sample Visualization

This section visualises **12 representative OCT images** (3 per class: CNV, DME, DRUSEN, NORMAL) in a clinical-style grid.  
Each panel shows:
- **True label** (top-left badge)
- **Predicted class** (bottom bar, ✓ green / ✗ red)
- **Confidence bar** for every class as a mini probability histogram

```python
import cv2
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.gridspec as gridspec
from matplotlib.patches import FancyBboxPatch

# - Configuration ─────────────────────────────────────────────────────────────
BASE_DIR   = '/content/Dataset - train+val+test/test'
CLASS_NAMES = ['CNV', 'DME', 'DRUSEN', 'NORMAL']
SAMPLES_PER_CLASS = 3          # 4 classes × 3 = 12 images
IMG_SIZE   = (128, 128)

# Hardcoded representative paths (3 per class) — matches the originals
SAMPLE_PATHS = [
    # CNV
    f'{BASE_DIR}/CNV/CNV-13823-1.jpeg',
    f'{BASE_DIR}/CNV/CNV-28682-8.jpeg',
    f'{BASE_DIR}/CNV/CNV-81630-7.jpeg',
    # DME
    f'{BASE_DIR}/DME/DME-30521-159.jpeg',
    f'{BASE_DIR}/DME/DME-30521-62.jpeg',
    f'{BASE_DIR}/DME/DME-30521-89.jpeg',
    # DRUSEN
    f'{BASE_DIR}/DRUSEN/DRUSEN-142234-6.jpeg',
    f'{BASE_DIR}/DRUSEN/DRUSEN-228939-3.jpeg',
    f'{BASE_DIR}/DRUSEN/DRUSEN-364469-18.jpeg',
    # NORMAL
    f'{BASE_DIR}/NORMAL/NORMAL-149928-10.jpeg',
    f'{BASE_DIR}/NORMAL/NORMAL-83787-8.jpeg',
    f'{BASE_DIR}/NORMAL/NORMAL-85367-4.jpeg',
]
TRUE_LABELS = ['CNV']*3 + ['DME']*3 + ['DRUSEN']*3 + ['NORMAL']*3

# - Colour palette ────────────────────────────────────────────────────────────
CLASS_COLORS = {
    'CNV':    '#E63946',   # crimson
    'DME':    '#457B9D',   # steel-blue
    'DRUSEN': '#E9C46A',   # amber
    'NORMAL': '#2A9D8F',   # teal
}
BG_COLOR     = '#0D1B2A'   # near-black navy
PANEL_COLOR  = '#1B2838'   # dark panel
TEXT_COLOR   = '#E8EDF2'   # off-white
CORRECT_CLR  = '#2A9D8F'
WRONG_CLR    = '#E63946'

# - Run inference on all 12 images ────────────────────────────────────────────
def predict_image(path):
    img_bgr = cv2.imread(path)
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
    img_resized = cv2.resize(img_rgb, IMG_SIZE)
    inp = np.array([img_resized], dtype=np.float32)
    probs = model.predict(inp, verbose=0)[0]   # shape (4,)
    pred_idx = np.argmax(probs)
    return img_rgb, probs, CLASS_NAMES[pred_idx]

results = [predict_image(p) for p in SAMPLE_PATHS]

# - Build the 4-row × 3-col figure ────────────────────────────────────────────
ROWS, COLS = 4, 3
fig = plt.figure(figsize=(18, 26), facecolor=BG_COLOR)

# Title
fig.text(0.5, 0.985, 'Retinal OCT — Diagnostic Model Review',
         ha='center', va='top', fontsize=20, fontweight='bold',
         color=TEXT_COLOR, fontfamily='monospace')
fig.text(0.5, 0.975, 'Model: eye_trained_model.keras  |  Classes: CNV · DME · DRUSEN · NORMAL',
         ha='center', va='top', fontsize=11, color='#7A9CC0', fontfamily='monospace')

# Legend bar at the top
legend_ax = fig.add_axes([0.06, 0.955, 0.88, 0.012])
legend_ax.set_xlim(0, 1)
legend_ax.set_ylim(0, 1)
legend_ax.axis('off')
legend_ax.set_facecolor(BG_COLOR)
for i, cn in enumerate(CLASS_NAMES):
    x = 0.12 + i * 0.22
    legend_ax.add_patch(FancyBboxPatch((x, 0.1), 0.08, 0.8,
                         boxstyle='round,pad=0.05',
                         facecolor=CLASS_COLORS[cn], edgecolor='none', alpha=0.9))
    legend_ax.text(x + 0.04, 0.5, cn, ha='center', va='center',
                   fontsize=10, fontweight='bold', color='white')

# Outer grid
outer = gridspec.GridSpec(ROWS, COLS, figure=fig,
                          left=0.04, right=0.96,
                          top=0.948, bottom=0.02,
                          hspace=0.30, wspace=0.12)

for idx in range(ROWS * COLS):
    row, col = divmod(idx, COLS)
    img_rgb, probs, pred_label = results[idx]
    true_label = TRUE_LABELS[idx]
    is_correct = (pred_label == true_label)
    accent = CORRECT_CLR if is_correct else WRONG_CLR
    check  = '✓' if is_correct else '✗'

    # Each panel = inner grid: image (top) + probability bar (bottom)
    inner = gridspec.GridSpecFromSubplotSpec(
        2, 1, subplot_spec=outer[row, col],
        height_ratios=[3.5, 1], hspace=0.06)

    # - Image axis ────────────────────────────────────────────────────────────
    ax_img = fig.add_subplot(inner[0])
    ax_img.imshow(img_rgb, cmap='gray')
    ax_img.set_xticks([]); ax_img.set_yticks([])
    for spine in ax_img.spines.values():
        spine.set_edgecolor(accent); spine.set_linewidth(2.5)
    ax_img.set_facecolor(PANEL_COLOR)

    # True-label badge (top-left)
    ax_img.text(0.03, 0.96, f' {true_label} ',
                transform=ax_img.transAxes,
                ha='left', va='top', fontsize=9, fontweight='bold',
                color='white',
                bbox=dict(facecolor=CLASS_COLORS[true_label],
                          edgecolor='none', boxstyle='round,pad=0.25',
                          alpha=0.92))

    # Sample index badge (top-right)
    ax_img.text(0.97, 0.96, f'#{idx+1:02d}',
                transform=ax_img.transAxes,
                ha='right', va='top', fontsize=8,
                color='#B0BEC5',
                bbox=dict(facecolor='#1B2838', edgecolor='none',
                          boxstyle='round,pad=0.2', alpha=0.75))

    # Prediction result (bottom of image)
    pred_conf = probs[CLASS_NAMES.index(pred_label)] * 100
    ax_img.text(0.5, 0.04,
                f'{check}  Pred: {pred_label}  ({pred_conf:.1f}%)',
                transform=ax_img.transAxes,
                ha='center', va='bottom', fontsize=9, fontweight='bold',
                color='white',
                bbox=dict(facecolor=accent, edgecolor='none',
                          boxstyle='round,pad=0.3', alpha=0.88))

    # - Probability bar axis ──────────────────────────────────────────────────
    ax_bar = fig.add_subplot(inner[1])
    ax_bar.set_facecolor(PANEL_COLOR)
    for spine in ax_bar.spines.values():
        spine.set_visible(False)

    bar_colors = [CLASS_COLORS[cn] for cn in CLASS_NAMES]
    bars = ax_bar.barh(CLASS_NAMES, probs * 100,
                        color=bar_colors, height=0.6,
                        edgecolor='none')

    # Highlight the predicted bar with a bright outline
    pred_pos = CLASS_NAMES.index(pred_label)
    bars[pred_pos].set_edgecolor(accent)
    bars[pred_pos].set_linewidth(1.8)

    ax_bar.set_xlim(0, 100)
    ax_bar.set_xticks([])
    ax_bar.tick_params(axis='y', labelsize=7.5, colors=TEXT_COLOR, pad=2)
    ax_bar.yaxis.set_tick_params(length=0)

    # Percentage labels on bars
    for bar, prob in zip(bars, probs):
        if prob * 100 > 3:
            ax_bar.text(bar.get_width() - 1, bar.get_y() + bar.get_height()/2,
                        f'{prob*100:.1f}%',
                        va='center', ha='right', fontsize=6.5,
                        color='white', fontweight='bold')

    ax_bar.set_facecolor(PANEL_COLOR)
    fig.patch.set_facecolor(BG_COLOR)

plt.savefig('medical_grid.png', dpi=150, bbox_inches='tight',
            facecolor=BG_COLOR, edgecolor='none')
plt.show()
print('Grid saved to medical_grid.png')
```

```python
# Add this import at the top (after existing imports)
import os
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns
from tensorflow.keras.preprocessing import image

# Add these functions after the imports section
def get_class_indices(data_dir):
    """Get class indices from directory structure"""
    classes = sorted(os.listdir(data_dir))
    class_indices = {class_name: idx for idx, class_name in enumerate(classes)}
    return class_indices, classes

def predict_single_image(img_path, model, img_size=(128, 128)):
    """Predict a single image and return top predictions"""
    img = image.load_img(img_path, target_size=img_size)
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0) / 255.0

    predictions = model.predict(img_array, verbose=0)[0]

    # Get top 2 predictions
    top_indices = np.argsort(predictions)[-2:][::-1]
    top_probs = predictions[top_indices]

    return top_indices, top_probs
```

```python
def evaluate_model_with_second_best(model, test_dir, class_indices, classes, img_size=(128, 128)):
    """Evaluate model and collect second best predictions"""
    results = []
    test_datagen = ImageDataGenerator(rescale=1./255)
    test_generator = test_datagen.flow_from_directory(
        test_dir,
        target_size=img_size,
        batch_size=32,
        class_mode='categorical',
        shuffle=False
    )

    # Get true labels
    true_classes = test_generator.classes

    # Predict
    predictions = model.predict(test_generator, verbose=1)

    # Analyze each prediction
    for i in range(len(predictions)):
        true_class_idx = true_classes[i]
        true_class_name = classes[true_class_idx]

        # Get top 2 predictions
        pred_indices = np.argsort(predictions[i])[-2:][::-1]
        pred_probs = predictions[i][pred_indices]

        pred_class_idx = pred_indices[0]
        pred_class_name = classes[pred_class_idx]
        pred_prob = pred_probs[0]

        second_best_idx = pred_indices[1]
        second_best_name = classes[second_best_idx]
        second_best_prob = pred_probs[1]

        is_correct = (pred_class_idx == true_class_idx)

        results.append({
            'true_class': true_class_name,
            'true_class_idx': true_class_idx,
            'predicted_class': pred_class_name,
            'predicted_prob': pred_prob,
            'predicted_class_idx': pred_class_idx,
            'second_best_class': second_best_name,
            'second_best_prob': second_best_prob,
            'second_best_idx': second_best_idx,
            'is_correct': is_correct,
            'confidence': pred_prob
        })

    return pd.DataFrame(results), predictions, true_classes
```

```python
def plot_top_classes_distribution(results_df, top_n=4):
    """Plot distribution of top classes and second best classes"""
    fig, axes = plt.subplots(2, 2, figsize=(20, 16))

    # 1. Top predicted classes
    top_pred_classes = results_df['predicted_class'].value_counts().head(top_n)
    axes[0, 0].barh(range(len(top_pred_classes)), top_pred_classes.values)
    axes[0, 0].set_yticks(range(len(top_pred_classes)))
    axes[0, 0].set_yticklabels(top_pred_classes.index)
    axes[0, 0].set_xlabel('Count')
    axes[0, 0].set_title(f'Top {top_n} Most Frequently Predicted Classes')
    axes[0, 0].invert_yaxis()

    # 2. Top second best classes
    top_second_classes = results_df['second_best_class'].value_counts().head(top_n)
    axes[0, 1].barh(range(len(top_second_classes)), top_second_classes.values)
    axes[0, 1].set_yticks(range(len(top_second_classes)))
    axes[0, 1].set_yticklabels(top_second_classes.index)
    axes[0, 1].set_xlabel('Count')
    axes[0, 1].set_title(f'Top {top_n} Most Frequent Second Best Predictions')
    axes[0, 1].invert_yaxis()

    # 3. Accuracy by true class
    accuracy_by_class = results_df.groupby('true_class').apply(
        lambda x: (x['is_correct'].sum() / len(x)) * 100
    ).sort_values(ascending=False).head(top_n)

    axes[1, 0].barh(range(len(accuracy_by_class)), accuracy_by_class.values)
    axes[1, 0].set_yticks(range(len(accuracy_by_class)))
    axes[1, 0].set_yticklabels(accuracy_by_class.index)
    axes[1, 0].set_xlabel('Accuracy (%)')
    axes[1, 0].set_title(f'Top {top_n} Most Accurate Classes')
    axes[1, 0].invert_yaxis()

    # 4. Confidence distribution for correct vs incorrect predictions
    correct_conf = results_df[results_df['is_correct']]['confidence']
    incorrect_conf = results_df[~results_df['is_correct']]['confidence']

    axes[1, 1].hist([correct_conf, incorrect_conf],
                   bins=20, alpha=0.7,
                   label=['Correct Predictions', 'Incorrect Predictions'])
    axes[1, 1].set_xlabel('Prediction Confidence')
    axes[1, 1].set_ylabel('Frequency')
    axes[1, 1].set_title('Confidence Distribution')
    axes[1, 1].legend()

    plt.tight_layout()
    plt.show()
```

```python
def plot_confusion_matrix_with_second_best(results_df, classes, top_n=4):
    """Plot confusion matrix focusing on most frequent classes"""
    from sklearn.metrics import confusion_matrix

    # Get most frequent true classes
    frequent_classes = results_df['true_class'].value_counts().head(top_n).index.tolist()
    class_indices = [classes.index(c) for c in frequent_classes]

    # Filter results
    filtered_results = results_df[results_df['true_class'].isin(frequent_classes)]

    # Create confusion matrix
    cm = confusion_matrix(filtered_results['true_class_idx'],
                         filtered_results['predicted_class_idx'],
                         labels=class_indices)

    plt.figure(figsize=(15, 12))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=frequent_classes,
                yticklabels=frequent_classes)
    plt.xlabel('Predicted Class')
    plt.ylabel('True Class')
    plt.title(f'Confusion Matrix (Top {top_n} Most Frequent Classes)')
    plt.xticks(rotation=45, ha='right')
    plt.yticks(rotation=0)
    plt.tight_layout()
    plt.show()
```

```python
def analyze_second_best_predictions(results_df):
    """Analyze when second best prediction is the true class"""
    second_best_correct = results_df[
        (results_df['second_best_idx'] == results_df['true_class_idx']) &
        (~results_df['is_correct'])
    ]

    print(f"\n{'='*60}")
    print("SECOND BEST PREDICTION ANALYSIS")
    print('='*60)
    print(f"Total samples: {len(results_df)}")
    print(f"Incorrect predictions: {len(results_df[~results_df['is_correct']])}")
    print(f"Cases where second best is correct: {len(second_best_correct)}")
    print(f"Percentage of incorrect where second best is correct: "
          f"{len(second_best_correct)/len(results_df[~results_df['is_correct']])*100:.2f}%")

    if len(second_best_correct) > 0:
        print("\nTop misclassified classes where second best was correct:")
        top_misclassified = second_best_correct['true_class'].value_counts().head(10)
        for class_name, count in top_misclassified.items():
            print(f"  {class_name}: {count} samples")

        print("\nConfidence comparison (when second best was correct):")
        print(f"  Average confidence in wrong prediction: {second_best_correct['predicted_prob'].mean():.3f}")
        print(f"  Average confidence in second best (true) prediction: {second_best_correct['second_best_prob'].mean():.3f}")
        print(f"  Average confidence difference: {abs(second_best_correct['predicted_prob'] - second_best_correct['second_best_prob']).mean():.3f}")
```

```python
# Main evaluation code (add at the end of your notebook)
# Set the test directory path
test_dir = '/content/Dataset - train+val+test/test/'

# Get class indices
class_indices, classes = get_class_indices(test_dir)
print(f"Found {len(classes)} classes:")
for i, class_name in enumerate(classes[:10]):  # Show first 10
    print(f"  {i}: {class_name}")
if len(classes) > 10:
    print(f"  ... and {len(classes)-10} more")

# Evaluate model
print("\nEvaluating model on test set...")
results_df, all_predictions, true_classes = evaluate_model_with_second_best(
    model, test_dir, class_indices, classes
)

# Calculate overall accuracy
accuracy = results_df['is_correct'].mean() * 100
print(f"\n{'='*60}")
print(f"MODEL EVALUATION RESULTS")
print('='*60)
print(f"Overall Accuracy: {accuracy:.2f}%")
print(f"Total Samples: {len(results_df)}")
print(f"Correct Predictions: {results_df['is_correct'].sum()}")
print(f"Incorrect Predictions: {len(results_df) - results_df['is_correct'].sum()}")

# Show sample predictions
print(f"\n{'='*60}")
print("SAMPLE PREDICTIONS (First 10 samples):")
print('='*60)
for i, row in results_df.head(10).iterrows():
    status = "✓ CORRECT" if row['is_correct'] else "✗ WRONG"
    print(f"\nSample {i}: {status}")
    print(f"  True Class: {row['true_class']}")
    print(f"  Predicted: {row['predicted_class']} ({row['predicted_prob']:.3f})")
    print(f"  Second Best: {row['second_best_class']} ({row['second_best_prob']:.3f})")

# Analyze second best predictions
analyze_second_best_predictions(results_df)

# Show misclassified examples
misclassified = results_df[~results_df['is_correct']]
if len(misclassified) > 0:
    print(f"\n{'='*60}")
    print("TOP MISCLASSIFICATION PATTERNS:")
    print('='*60)

    # Most common misclassifications
    misclassification_patterns = misclassified.groupby(
        ['true_class', 'predicted_class']
    ).size().sort_values(ascending=False).head(10)

    for (true_class, pred_class), count in misclassification_patterns.items():
        print(f"{true_class} → {pred_class}: {count} samples")

# Generate classification report
y_true = results_df['true_class_idx']
y_pred = results_df['predicted_class_idx']
print(f"\n{'='*60}")
print("CLASSIFICATION REPORT:")
print('='*60)
print(classification_report(y_true, y_pred, target_names=classes, zero_division=0))

# Plot visualizations
print(f"\n{'='*60}")
print("VISUALIZATIONS:")
print('='*60)

# Plot top classes distribution
plot_top_classes_distribution(results_df, top_n=15)

# Plot confusion matrix (top 20 classes for readability)
plot_confusion_matrix_with_second_best(results_df, classes, top_n=20)

# Additional analysis: Confidence by class
print(f"\n{'='*60}")
print("CONFIDENCE ANALYSIS BY CLASS:")
print('='*60)
confidence_by_class = results_df.groupby('true_class').agg({
    'confidence': ['mean', 'std', 'count'],
    'is_correct': 'mean'
}).round(3)

confidence_by_class.columns = ['avg_confidence', 'std_confidence', 'sample_count', 'accuracy']
confidence_by_class = confidence_by_class.sort_values('accuracy', ascending=False)

print(confidence_by_class.head(10))
```

```python
def triangle_plot(x, y, xlabel, ylabel, panel):
    plt.scatter(x, y, s=20, alpha=0.85)

    # Triangle boundaries
    plt.plot([0, 1], [0, 0], linestyle='--')   # bottom
    # plt.plot([0, 1], [1, 0], linestyle='--')   # hypotenuse
    x = np.linspace(0, 1, 100)
    plt.plot(x, 1 - x, linestyle='--')

    plt.plot([0, 1], [0, 1], linestyle='--')   # diagonal reference

    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.text(0.03, 0.94, panel, fontsize=14, fontweight='bold')
```

```python
import numpy as np
import matplotlib.pyplot as plt

def prepData(DATASET):
  # Get class names safely
  class_name = DATASET.class_names
  num_classes = len(class_name)

  print("Number of classes:", num_classes)
  # --- Get predictions for validation set ---
  y_prob = model.predict(DATASET)

  # True labels (one-hot -> index)
  # y_true = np.concatenate([y.numpy() for _, y in test_set], axis=0)
  y_true = np.concatenate([y for _, y in DATASET], axis=0)

  y_true_idx = np.argmax(y_true, axis=1)

  # Top-1 prediction
  top1_idx = np.argmax(y_prob, axis=1)

  # Second-best prediction
  top2_idx = np.argsort(y_prob, axis=1)[:, -2]

  # Map to class names
  y_true_names = [class_name[i] for i in y_true_idx]
  top1_names = [class_name[i] for i in top1_idx]
  top2_names = [class_name[i] for i in top2_idx]

  top1_correct = (y_true_idx == top1_idx)
  top2_correct = (y_true_idx == top2_idx)

  top1_accuracy = np.mean(top1_correct)
  # top2_accuracy = np.mean(top1_correct | top2_correct)
  top2_pred = np.argsort(y_prob, axis=1)[:, -2:]
  top2_accuracy = np.mean([
    y_true_idx[i] in top2_pred[i]
    for i in range(len(y_true_idx))
  ])

  print(f"Top-1 Accuracy: {top1_accuracy:.4f}")
  print(f"Top-2 Accuracy: {top2_accuracy:.4f}")

  labels = ["Top-1 correct", "Only Top-2 correct", "Not in Top-2"]

  top1_only = top1_correct
  top2_only = (~top1_correct) & top2_correct
  not_in_top2 = (~top1_correct) & (~top2_correct)

  counts = [
    np.sum(top1_only),
    np.sum(top2_only),
    np.sum(not_in_top2)
  ]

  # Sort probabilities per sample
  sorted_probs = np.sort(y_prob, axis=1)

  # C_best = sorted_probs[:, -1]
  C_best = np.max(y_prob, axis=1)

  C_second = sorted_probs[:, -2]
  C_true = y_prob[np.arange(len(y_true_idx)), y_true_idx]

  print("Quick debug check")
  print("Any C_true > C_best:",
      np.any(C_true > C_best))

  print("Max C_best:", np.max(C_best))
  print("Min C_best:", np.min(C_best))

  print("Check probabilities sum to 1:",
      np.allclose(np.sum(y_prob, axis=1), 1, atol=1e-3))

  plt.figure(figsize=(12, 4))

  # (a) C_true vs C_secondBest
  plt.subplot(1, 3, 1)
  triangle_plot(
    x=C_true,
    y=C_second,
    xlabel=r"$C_{True}$",
    ylabel=r"$C_{SecondBest}$",
    panel="a"
  )

  # (b) C_best vs C_secondBest
  plt.subplot(1, 3, 2)
  triangle_plot(
    x=C_best,
    y=C_second,
    xlabel=r"$C_{Best}$",
    ylabel=r"$C_{SecondBest}$",
    panel="b"
  )

  plt.subplot(1, 3, 3)
  triangle_plot(
    x=C_best,
    y=C_true,
    xlabel=r"$C_{Best}$",
    ylabel=r"$C_{True}$",
    panel="c"
  )

  plt.tight_layout()
  plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import os

def plot_metrics_separately(labels, angles, metrics):
    """Plots each metric in a separate polar plot with premium styling."""
    colors = {'Precision': '#4a627a', 'Recall': '#50a18f', 'F1-score': '#ff8c7a'}
    markers = {'Precision': 'o', 'Recall': 's', 'F1-score': '^'}

    for title, values in metrics.items():
        vals = values + values[:1]
        fig = plt.figure(figsize=(10, 10), facecolor='#f8faff')
        ax = plt.subplot(111, polar=True)
        ax.set_facecolor('#fdfdff')

        color = colors.get(title, '#4a627a')
        marker = markers.get(title, 'o')

        ax.plot(angles, vals, linewidth=3, linestyle='-', label=title, color=color, marker=marker, markersize=8, markerfacecolor='white', markeredgewidth=2)
        ax.fill(angles, vals, color=color, alpha=0.15)

        ax.set_thetagrids(np.degrees(angles[:-1]), labels, fontsize=11, fontweight='bold', color='#333333')
        ax.set_ylim(0, 1)
        ax.grid(True, linestyle='--', alpha=0.6, color='#cbd5e0')

        ax.set_title(f"Class-wise {title}", size=24, weight='bold', color='#1a365d', y=1.1, pad=30)

        # Legend styling
        legend = ax.legend(loc='upper right', bbox_to_anchor=(1.25, 1.1), frameon=True, shadow=True)
        legend.get_frame().set_facecolor('white')
        legend.get_frame().set_edgecolor('#1a365d')

        plt.tight_layout()
        plt.show()

def plot_all_metrics_combined(labels, angles, metrics):
    """Plots all metrics on the same polar plot with premium styling."""
    fig = plt.figure(figsize=(12, 12), facecolor='#f8faff')
    ax = plt.subplot(111, polar=True)
    ax.set_facecolor('#fdfdff')

    # Premium colors from reference image
    colors = ['#4a627a', '#50a18f', '#ff2211'] # Deep Blue, Teal, Coral
    markers = ['o', 's', '^'] # circle, square, triangle

    for (title, values), color, marker in zip(metrics.items(), colors, markers):
        vals = values + values[:1]
        ax.plot(angles, vals, linewidth=3, linestyle='-', label=title, color=color, marker=marker, markersize=8, markerfacecolor='white', markeredgewidth=2)
        ax.fill(angles, vals, color=color, alpha=0.08)

    ax.set_thetagrids(np.degrees(angles[:-1]), labels, fontsize=11, fontweight='bold', color='#333333')
    ax.set_ylim(0, 1)
    ax.grid(True, linestyle='--', alpha=0.6, color='#cbd5e0')

    # Title styling matching the reference header
    plt.suptitle("Model Evaluation: Classification Metrics", size=28, weight='bold', color='#1a365d', y=0.98)
    ax.set_title("Per-class performance comparison", size=16, style='italic', color='#4a627a', pad=40)

    # Legend styling
    legend = ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1), frameon=True, shadow=True, fontsize=12)
    legend.get_frame().set_facecolor('white')
    legend.get_frame().set_edgecolor('#1a365d')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95])
    plt.show()

def radarPlot():

    csv_path = ('./Classification.csv')

    if not os.path.exists(csv_path):
        # Fallback if __file__ is not available (e.g. running in some interactive environments)
        csv_path = 'Classification.csv'

    df = pd.read_csv(csv_path)

    # Filter out average/accuracy rows
    df_filtered = df[~df['class'].astype(str).isin(['accuracy', 'macro avg', 'weighted avg'])].copy()

    labels = df_filtered['class'].astype(str).tolist()
    metrics = {
        "Precision": df_filtered['precision'].tolist(),
        "Recall": df_filtered['recall'].tolist(),
        "F1-score": df_filtered['f1_score'].tolist()
    }

    angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False).tolist()
    angles += angles[:1]

    # Execute plotting
    #print("Generating separate plots...")
    #plot_metrics_separately(labels, angles, metrics)

    print("Generating combined plot...")
    plot_all_metrics_combined(labels, angles, metrics)


radarPlot()
```
